```python
l = [1,2,3,4,5,10.4,True,"hello","Chinna",10+5j]
```

```python
l_even=[]
l_odd=[]
l_bool=[]
l_str=[]
l_complex=[]
l_float=[]
for i in l:
  if type(i)==int:
    if i%2==0:
      l_even.append(i)
    else:
      l_odd.append(i)
  elif type(i)==bool:
    l_bool.append(i)
  elif type(i)==str:
    l_str.append(i)
  elif type(i)==complex:
    l_complex.append(i)
  else:
    l_float.append(i)
```

```python
l_even
```

```python
l_odd
```

```python
l_str
```

```python
l_bool
```

```python
l_complex
```

```python
l_float
```
